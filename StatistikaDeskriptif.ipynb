```python
# =========================================================
# TUGAS CASE - STATISTIKA DESKRIPTIF
# Nama       : Nazwa Azzahra
# NIM        : 20240040230
# Kelas      : TI 24 F
# Mata Kuliah: Data Science (Sesi Pertemuan 2)
# Dosen      : Alun Sujjada, S.Kom., M.T.
# =========================================================

import numpy as np
import pandas as pd

print("="*60)
print("  PEMPROSESAN TUGAS STATISTIKA DESKRIPTIF - DATA SCIENCE  ")
print("="*60)

# ---------------------------------------------------------
# SOAL 1: PERHITUNGAN DATA GAJI KARYAWAN
# ---------------------------------------------------------
gaji = [3, 4, 4, 5, 5, 6, 6, 7, 7, 7, 8, 8, 9, 9, 10, 10, 12, 15, 20, 25]

mean_gaji = np.mean(gaji)
median_gaji = np.median(gaji)
modus_gaji = pd.Series(gaji).mode()[0]
range_gaji = np.ptp(gaji)

print("\n[SOAL 1] Data Gaji Karyawan (Juta Rupiah)")
print(f"Data Raw       : {gaji}")
print(f"a. Mean        : {mean_gaji:.2f} Juta")
print(f"   Median      : {median_gaji:.2f} Juta")
print(f"   Modus       : {modus_gaji} Juta")
print(f"   Range       : {range_gaji} Juta")

# ---------------------------------------------------------
# SOAL 4: VARIANS DAN SIMPANGAN BAKU
# ---------------------------------------------------------
nilai = [60, 65, 70, 75, 75, 80, 80, 85, 90, 95]

mean_nilai = np.mean(nilai)
varians_sampel = np.var(nilai, ddof=1)  # ddof=1 untuk sampel
std_sampel = np.std(nilai, ddof=1)

print("\n" + "-"*60)
print("[SOAL 4] Analisis Variansi Nilai Ujian 10 Mahasiswa")
print(f"Data Nilai     : {nilai}")
print(f"Rata-rata (Mean): {mean_nilai:.2f}")
print(f"Varians Sampel  : {varians_sampel:.2f}")
print(f"Simpangan Baku : {std_sampel:.2f}")
print("="*60)
```
